````python
increase_amount_cost = 500

collection_order.update_many(
    {"items_id": item_id_obj},
    {
        "$push": {"items_id": item_id_obj},
        "$inc": {"total_sum": increase_amount_cost}
    }
)

print("Orders updated with new item and increased total cost.")
````
