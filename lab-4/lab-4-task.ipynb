6. Напишіть запити, які вибирають товари за різними критеріям і їх сукупності: 
- категорія та ціна (в проміжку) - конструкція \$and, 
- модель чи одна чи інша -конструкція \$or,
- виробники з переліку - конструкція \$in


```python
min = 500
max = 3000

cat_and_price = collection_item.find( { "$and": [ { "category": cat }, { "price": {"$gte": min, "$lte": max}} ] } )
print(f'Товар у категорії {cat} та ціновому діапазоні від {min}  до {max}: {json.dumps([item for item in cat_and_price], indent=4, default=str)}')
```

```python
model_1 = "MacBook Air 10"
model_2 = "Sony I32"

model_one_more = collection_item.find( { "$or": [ { "model": model_1 }, { "model": model_2 } ] })
print(f'Товар за моделями {model_1} або  {model_2} : {json.dumps([item for item in model_one_more], indent=4, default=str)}')
```

```python
producer_list = ["Sony", "Apple" ]

producer_from_list = collection_item.find( { "producer" : {"$in": producer_list}})
print(f'Виробник з переліку {producer_list}: {json.dumps([item for item in producer_from_list], indent=4, default=str)}')
```
